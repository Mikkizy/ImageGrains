```python
def weighted_crossentropy(y_true, y_pred):
    """
    Calculates the weighted cross-entropy loss between the true labels and predicted labels.

    Parameters
    ----------
    y_true : tensor
        True labels.
    y_pred : tensor
        Predicted labels.

    Returns
    -------
    loss : tensor
        Weighted cross-entropy loss.
    """
    class_weights = tf.constant(
        [[[[0.6, 1.0, 5.0]]]]
    )  # increase the weight on the grains and the grain boundaries
    unweighted_losses = tf.nn.softmax_cross_entropy_with_logits(
        labels=y_true, logits=y_pred
    )
    weights = tf.reduce_sum(class_weights * y_true, axis=-1)
    weighted_losses = weights * unweighted_losses
    loss = tf.reduce_mean(weighted_losses)
    return loss
```

```python
def predict_image_tile_tflite(im_tile, interpreter):
    """
    Predicts one image tile using a TFLite model.

    Parameters
    ----------
    im_tile : 3D array
        The image tile for which the prediction will be done.
    interpreter
        TFLite interpreter for the segmentation model.

    Returns
    -------
    im_tile_pred : 3D array
        Predicted tile.
    """
    # Check for invalid input
    if not isinstance(im_tile, np.ndarray):
        raise ValueError("Input image tile must be a numpy array.")
    if len(im_tile.shape) != 3:
        raise ValueError("Input image tile must be a 3D array.")
    if len(im_tile.shape) == 3 and im_tile.shape[2] != 3:
        raise ValueError("3D input image tile must have 3 channels.")

    # Get input and output details
    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()

    # Add batch dimension
    im_tile = np.expand_dims(im_tile, axis=0)

    # Check if input type conversion is needed for quantized model
    if input_details[0]['dtype'] == np.uint8 or input_details[0]['dtype'] == np.int8:
        # Quantize the input
        scale, zero_point = input_details[0]['quantization']
        im_tile_quantized = (im_tile / scale + zero_point).astype(input_details[0]['dtype'])
        # Set the tensor
        interpreter.set_tensor(input_details[0]['index'], im_tile_quantized)
    else:
        # For float models, just set the tensor directly
        interpreter.set_tensor(input_details[0]['index'], im_tile.astype(np.float32))

    # Run inference
    interpreter.invoke()

    # Get the output tensor
    im_tile_pred = interpreter.get_tensor(output_details[0]['index'])

    # Dequantize the output if needed
    if output_details[0]['dtype'] == np.uint8 or output_details[0]['dtype'] == np.int8:
        scale, zero_point = output_details[0]['quantization']
        im_tile_pred = scale * (im_tile_pred.astype(np.float32) - zero_point)

    # Remove batch dimension
    im_tile_pred = im_tile_pred[0]

    return im_tile_pred
```

```python
from tqdm import trange
def predict_image_tflite(image, interpreter, I):
    """
    Semantic segmentation of the entire image using a TFLite model.

    Parameters
    ----------
    image : 2D or 3D array
        The image that is being segmented. Can have one or more channels.
    interpreter
        TFLite interpreter for the segmentation model.
    I : int
        Size of the square-shaped image tiles in pixels.

    Returns
    -------
    image_pred : 3D array
        Semantic segmentation result for the input image.
    """
    # Check for invalid input
    if not isinstance(image, np.ndarray):
        raise ValueError("Input image must be a numpy array.")
    if image.ndim not in [2, 3]:
        raise ValueError("Input image must be a 2D or 3D array.")
    if image.ndim == 3 and image.shape[2] not in [1, 3]:
        raise ValueError("3D input image must have 1 or 3 channels.")

    pad_rows = I - np.mod(image.shape[0], I)
    pad_cols = I - np.mod(image.shape[1], I)
    if image.ndim == 2:
        image = np.stack((image, image, image), axis=-1)  # convert to 3 channels
    if image.ndim == 3:
        image = np.vstack((image, np.zeros((pad_rows, image.shape[1], image.shape[2]))))
        image = np.hstack((image, np.zeros((image.shape[0], pad_cols, image.shape[2]))))
    r = int(np.floor(image.shape[0] / I))  # number of rows of image tiles
    c = int(np.floor(image.shape[1] / I))  # number of columns of image tiles

    I2 = int(I / 2)
    W = np.hanning(I) * np.hanning(I)[:, np.newaxis]
    Wup = W.copy()
    Wup[:I2, :] = np.tile(np.hanning(I), (I2, 1))
    Wdown = W.copy()
    Wdown[I2:, :] = np.tile(np.hanning(I), (I2, 1))

    image = np.hstack(
        (np.zeros((r * I, I2, 3)), image, np.zeros((r * I, I2, 3)))
    )  # padding on the left and right sides
    image_pred = np.zeros((image.shape[0], image.shape[1], 3))
    print("segmenting image tiles...")
    for i in trange(c + 1):  # rows, no offset
        for j in range(1, 2 * r - 2):  # columns
            im_tile = image[j * I2 : (j + 2) * I2, i * I : (i + 1) * I, :] / 255.0
            im_tile_pred = predict_image_tile_tflite(im_tile, interpreter)
            for layer in range(3):
                image_pred[j * I2 : (j + 2) * I2, i * I : (i + 1) * I, layer] += (
                    im_tile_pred[:, :, layer] * W
                )
    for i in range(c + 1):  # first row
        im_tile = image[: 2 * I2, i * I : (i + 1) * I, :] / 255.0
        im_tile_pred = predict_image_tile_tflite(im_tile, interpreter)
        for layer in range(3):
            image_pred[: 2 * I2, i * I : (i + 1) * I, layer] += (
                im_tile_pred[:, :, layer] * Wup
            )
    for i in range(c + 1):  # last row
        im_tile = image[(2 * r - 2) * I2 : 2 * r * I2, i * I : (i + 1) * I, :] / 255.0
        im_tile_pred = predict_image_tile_tflite(im_tile, interpreter)
        for layer in range(3):
            image_pred[(2 * r - 2) * I2 : 2 * r * I2, i * I : (i + 1) * I, layer] += (
                im_tile_pred[:, :, layer] * Wdown
            )
    for i in trange(c):  # rows, half offset
        for j in range(1, 2 * r - 2):  # columns
            im_tile = (
                image[j * I2 : (j + 2) * I2, i * I + I2 : (i + 1) * I + I2, :] / 255.0
            )
            im_tile_pred = predict_image_tile_tflite(im_tile, interpreter)
            for layer in range(3):
                image_pred[
                    j * I2 : (j + 2) * I2, i * I + I2 : (i + 1) * I + I2, layer
                ] += (im_tile_pred[:, :, layer] * W)
    for i in range(c):  # first row
        im_tile = image[: 2 * I2, i * I + I2 : (i + 1) * I + I2, :] / 255.0
        im_tile_pred = predict_image_tile_tflite(im_tile, interpreter)
        for layer in range(3):
            image_pred[: 2 * I2, i * I + I2 : (i + 1) * I + I2, layer] += (
                im_tile_pred[:, :, layer] * Wup
            )
    for i in range(c):  # last row
        im_tile = (
            image[(2 * r - 2) * I2 : 2 * r * I2, i * I + I2 : (i + 1) * I + I2, :]
            / 255.0
        )
        im_tile_pred = predict_image_tile_tflite(im_tile, interpreter)
        for layer in range(3):
            image_pred[
                (2 * r - 2) * I2 : 2 * r * I2, i * I + I2 : (i + 1) * I + I2, layer
            ] += (im_tile_pred[:, :, layer] * Wdown)

    image_pred = image_pred[:, I2:-I2, :]  # crop the left and right side padding
    image_pred = image_pred[:-pad_rows, :-pad_cols, :]  # get rid of padding
    return image_pred
```

```python
import numpy as np
from PIL import Image
import io
import pathlib

def load_image_for_tflite(
    path,
    color_mode="rgb",
):
    """
    Loads an image as a NumPy array, ready for TFLite model input.

    Parameters
    ----------
    path : str, pathlib.Path, or io.BytesIO
        Path to the image file or BytesIO object containing image data.
    color_mode : str
        One of "grayscale" or "rgb". Default: "rgb".
        The desired image color format.

    Returns
    -------
    numpy.ndarray
        The loaded image as a NumPy array with shape (height, width, channels)
        or (height, width) for grayscale.
    """
    # Load the image using PIL
    if isinstance(path, io.BytesIO):
        img = Image.open(path)
    elif isinstance(path, (pathlib.Path, bytes, str)):
        if isinstance(path, pathlib.Path):
            path = str(path.resolve())
        img = Image.open(path)
    else:
        raise TypeError(f"path should be path-like or io.BytesIO, not {type(path)}")

    # Convert to the desired color mode
    if color_mode == "grayscale":
        if img.mode != "L":
            img = img.convert("L")
    elif color_mode == "rgb":
        if img.mode != "RGB":
            img = img.convert("RGB")
    else:
        raise ValueError('color_mode must be "grayscale" or "rgb"')

    # Convert to NumPy array
    img_array = np.array(img)

    return img_array
```

```python
import torch
import numpy as np
from mobile_sam import sam_model_registry
import onnx
import onnxruntime as ort

def export_mobilesam_encoder(pytorch_model_path="./mobile_sam.pt",
                            encoder_onnx_path="./mobile_sam_encoder.onnx"):
    """
    Export only the image encoder from MobileSAM to ONNX for Android.
    """
    print("🔄 Exporting MobileSAM image encoder...")

    try:
        # Load the PyTorch MobileSAM model
        sam_model = sam_model_registry["vit_t"](checkpoint=pytorch_model_path)
        sam_model.eval()

        # Extract the image encoder
        image_encoder = sam_model.image_encoder

        print(f"Image encoder type: {type(image_encoder)}")

        # Create dummy input (1024x1024 RGB image)
        dummy_input = torch.randn(1, 3, 1024, 1024)

        # Test the encoder first
        with torch.no_grad():
            test_output = image_encoder(dummy_input)
            print(f"Encoder output shape: {test_output.shape}")

        # Export to ONNX
        torch.onnx.export(
            image_encoder,
            dummy_input,
            encoder_onnx_path,
            export_params=True,
            opset_version=17,  # Use newer opset for better mobile support
            do_constant_folding=True,
            input_names=["input_image"],
            output_names=["image_embeddings"],
            dynamic_axes={
                "input_image": {0: "batch_size"},
                "image_embeddings": {0: "batch_size"}
            }
        )

        print(f"✅ Image encoder exported: {encoder_onnx_path}")

        # Verify the ONNX model
        onnx_model = onnx.load(encoder_onnx_path)
        onnx.checker.check_model(onnx_model)
        print("✅ ONNX encoder verification passed!")

        # Test with ONNX Runtime
        session = ort.InferenceSession(encoder_onnx_path)

        # Test inference
        dummy_input_np = dummy_input.numpy()
        onnx_output = session.run(["image_embeddings"], {"input_image": dummy_input_np})

        print(f"ONNX encoder output shape: {onnx_output[0].shape}")

        # Compare outputs
        torch_output_np = test_output.numpy()
        max_diff = np.abs(torch_output_np - onnx_output[0]).max()
        print(f"Max difference between PyTorch and ONNX: {max_diff:.8f}")

        # Get file size
        import os
        file_size = os.path.getsize(encoder_onnx_path) / (1024 * 1024)
        print(f"Encoder ONNX size: {file_size:.2f} MB")

        return encoder_onnx_path

    except Exception as e:
        print(f"❌ Failed to export encoder: {e}")
        return None
```

```python
import onnxruntime as ort
import numpy as np
import cv2
from tqdm import tqdm
from skimage import measure
from skimage.measure import regionprops_table
import pandas as pd
import matplotlib.pyplot as plt

class PureONNXMobileSAMFixed:
    """
    Pure ONNX MobileSAM implementation with proper coordinate transformation
    based on the official SAM coordinate transform logic.
    """

    def __init__(self, encoder_onnx_path, predictor_onnx_path, device='cpu'):
        self.device = device
        self.encoder_onnx_path = encoder_onnx_path
        self.predictor_onnx_path = predictor_onnx_path

        # ONNX sessions
        self.encoder_session = None
        self.predictor_session = None

        # Image processing info
        self.cached_embeddings = None
        self.cached_image_hash = None
        self.original_image_shape = None
        self.target_length = 1024  # SAM's standard

        print(f"Initializing Pure ONNX MobileSAM:")
        print(f"  Encoder: {encoder_onnx_path}")
        print(f"  Predictor: {predictor_onnx_path}")

    def load_models(self):
        """Load both ONNX models."""
        try:
            providers = ['CUDAExecutionProvider', 'CPUExecutionProvider'] if self.device == 'cuda' else ['CPUExecutionProvider']

            self.encoder_session = ort.InferenceSession(self.encoder_onnx_path, providers=providers)
            self.predictor_session = ort.InferenceSession(self.predictor_onnx_path, providers=providers)

            print("✅ Both ONNX models loaded successfully")
            return True

        except Exception as e:
            print(f"❌ Failed to load ONNX models: {e}")
            return False

    def apply_coords_transform(self, coords, original_size):
        """
        Apply SAM's coordinate transformation logic.
        This replicates predictor.transform.apply_coords() functionality.
        """
        # Get the original image dimensions
        old_h, old_w = original_size

        # Calculate the scale to fit the longer side to target_length (1024)
        scale = self.target_length * 1.0 / max(old_h, old_w)

        # Calculate new dimensions after scaling
        new_h = int(old_h * scale)
        new_w = int(old_w * scale)

        # Apply the scale to coordinates
        coords_scaled = coords * scale

        print(f"Coordinate transform: {original_size} -> scale={scale:.4f} -> new_size=({new_h}, {new_w})")

        return coords_scaled.astype(np.float32)

    def preprocess_image_sam_style(self, image):
        """
        Preprocess image exactly like SAM does internally.
        """
        # Store original shape
        self.original_image_shape = image.shape[:2]  # (height, width)
        old_h, old_w = self.original_image_shape

        # Convert BGR to RGB if needed
        if len(image.shape) == 3 and image.shape[2] == 3:
            rgb_image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        else:
            rgb_image = image.copy()

        # Calculate scale to fit longer side to 1024
        scale = self.target_length * 1.0 / max(old_h, old_w)
        new_h = int(old_h * scale)
        new_w = int(old_w * scale)

        print(f"SAM preprocessing: {old_w}x{old_h} -> scale={scale:.4f} -> {new_w}x{new_h}")

        # Resize maintaining aspect ratio
        resized = cv2.resize(rgb_image, (new_w, new_h), interpolation=cv2.INTER_LINEAR)

        # Pad to 1024x1024 (SAM's standard input size)
        # Create 1024x1024 canvas
        padded = np.zeros((self.target_length, self.target_length, 3), dtype=np.uint8)

        # Place resized image in top-left corner
        padded[:new_h, :new_w] = resized

        # Normalize using SAM's normalization
        normalized = padded.astype(np.float32) / 255.0
        mean = np.array([0.485, 0.456, 0.406], dtype=np.float32)
        std = np.array([0.229, 0.224, 0.225], dtype=np.float32)
        normalized = (normalized - mean.reshape(1, 1, 3)) / std.reshape(1, 1, 3)

        # Convert to tensor format
        input_tensor = np.transpose(normalized, (2, 0, 1)).astype(np.float32)
        input_tensor = np.expand_dims(input_tensor, axis=0).astype(np.float32)

        return input_tensor

    def get_image_embeddings(self, image):
        """Get image embeddings with proper SAM preprocessing."""
        # Check cache
        image_hash = hash(image.data.tobytes())

        if (self.cached_embeddings is not None and
            self.cached_image_hash == image_hash):
            print("Using cached image embeddings")
            return self.cached_embeddings

        print("Generating image embeddings with SAM-style preprocessing...")

        try:
            # Preprocess with SAM's exact method
            input_tensor = self.preprocess_image_sam_style(image)

            # Run encoder
            embeddings = self.encoder_session.run(
                ["image_embeddings"],
                {"input_image": input_tensor}
            )[0]

            # Cache embeddings
            self.cached_embeddings = embeddings.astype(np.float32)
            self.cached_image_hash = image_hash

            print(f"✅ Generated embeddings shape: {embeddings.shape}")
            return self.cached_embeddings

        except Exception as e:
            print(f"❌ Failed to get embeddings: {e}")
            return None

    def predict_mask_with_sam_coords(self, x, y, image_embeddings):
        """
        Predict mask using proper SAM coordinate transformation.
        Replicates the exact logic from the sample code.
        """
        try:
            if self.original_image_shape is None:
                print("❌ Original image shape not set. Run get_image_embeddings first.")
                return None

            # Prepare point input exactly like the sample code
            input_point = np.array([[x, y]])  # Shape: (1, 2)
            input_label = np.array([1])  # Positive point

            # Add batch index and padding point (from sample code)
            onnx_coord = np.concatenate([input_point, np.array([[0.0, 0.0]])], axis=0)[None, :, :]
            onnx_label = np.concatenate([input_label, np.array([-1])], axis=0)[None, :].astype(np.float32)

            # CRITICAL: Apply SAM's coordinate transformation
            onnx_coord = self.apply_coords_transform(onnx_coord, self.original_image_shape)

            # Prepare other inputs exactly like sample code
            onnx_mask_input = np.zeros((1, 1, 256, 256), dtype=np.float32)
            onnx_has_mask_input = np.zeros(1, dtype=np.float32)

            # Package inputs exactly like sample code
            ort_inputs = {
                "image_embeddings": image_embeddings,
                "point_coords": onnx_coord,
                "point_labels": onnx_label,
                "mask_input": onnx_mask_input,
                "has_mask_input": onnx_has_mask_input,
                "orig_im_size": np.array(self.original_image_shape, dtype=np.float32)  # [height, width]
            }

            # Run ONNX inference
            masks, _, _ = self.predictor_session.run(None, ort_inputs)

            # Threshold using SAM's default (0.0, not 0.5!)
            binary_masks = masks > 0.0

            # Take the first mask
            mask = binary_masks[0, 0]  # Remove batch and mask index dimensions

            return mask

        except Exception as e:
            print(f"❌ SAM coordinate prediction failed for point ({x}, {y}): {e}")
            import traceback
            traceback.print_exc()
            return None


def pure_onnx_fixed_mobilesam_segmentation(
    encoder_onnx_path,
    predictor_onnx_path,
    image,
    image_pred,
    coords,
    labels,
    min_area,
    plot_image=False,
    remove_edge_grains=False,
):
    """
    Pure ONNX MobileSAM segmentation with proper SAM coordinate handling.
    """
    from segmenteverygrain.segmenteverygrain import (
        collect_polygon_from_mask, find_connected_components,
        merge_overlapping_polygons, create_labeled_image,
        plot_image_w_colorful_grains, plot_grain_axes_and_centroids
    )

    print(f"Pure ONNX Fixed MobileSAM segmentation with {len(coords)} coordinates...")

    # Initialize pure ONNX MobileSAM
    mobilesam = PureONNXMobileSAMFixed(encoder_onnx_path, predictor_onnx_path)

    if not mobilesam.load_models():
        print("Failed to load Pure ONNX MobileSAM models")
        return [], np.zeros_like(image[:, :, 0]), np.zeros_like(image), pd.DataFrame(), None, None

    # Get image embeddings with proper SAM preprocessing
    print("Getting image embeddings...")
    image_embeddings = mobilesam.get_image_embeddings(image)
    if image_embeddings is None:
        print("Failed to get image embeddings")
        return [], np.zeros_like(image[:, :, 0]), np.zeros_like(image), pd.DataFrame(), None, None

    # Process coordinates with SAM coordinate transforms
    all_grains = []
    successful_coords = 0

    for i in tqdm(range(len(coords)), desc="Processing with SAM coordinate transforms"):
        x, y = coords[i, 0], coords[i, 1]

        # Use SAM-style coordinate transformation
        mask = mobilesam.predict_mask_with_sam_coords(x, y, image_embeddings)

        if mask is not None and np.sum(mask) > 0:
            # Handle multiple connected components (same as existing)
            temp_labels, n_elems = measure.label(mask, return_num=True, connectivity=1)
            if n_elems > 1:
                click_point = np.array([int(y), int(x)])
                min_distance = float('inf')
                best_label = 1

                for label_id in range(1, n_elems + 1):
                    component_coords = np.where(temp_labels == label_id)
                    if len(component_coords[0]) > 0:
                        component_points = np.column_stack((component_coords[0], component_coords[1]))
                        distances = np.linalg.norm(component_points - click_point, axis=1)
                        min_dist_to_component = np.min(distances)
                        if min_dist_to_component < min_distance:
                            min_distance = min_dist_to_component
                            best_label = label_id

                mask = (temp_labels == best_label)

            # Process grain (same as existing)
            contours = measure.find_contours(mask.astype(float), 0.5)
            if contours:
                largest_contour = max(contours, key=len)
                sx = largest_contour[:, 1]
                sy = largest_contour[:, 0]

                # Handle edge cases
                if (np.any(mask[0, :]) or np.any(mask[-1, :]) or
                    np.any(mask[:, 0]) or np.any(mask[:, -1])):
                    mask = np.pad(mask, 1, mode="constant")
                    contours = measure.find_contours(mask.astype(float), 0.5)
                    if contours:
                        largest_contour = max(contours, key=len)
                        sx = largest_contour[:, 1]
                        sy = largest_contour[:, 0]
                        if np.any(mask[1, :]):
                            sy = sy - 1
                        if np.any(mask[:, 1]):
                            sx = sx - 1
                    mask = mask[1:-1, 1:-1]

                # Collect grain
                if np.max(mask) > 0:
                    successful_coords += 1
                    if (remove_edge_grains and
                        np.sum(np.hstack([mask[:4, :], mask[-4:, :],
                                        mask[:, :4].T, mask[:, -4:].T])) == 0):
                        all_grains = collect_polygon_from_mask(labels, mask, image_pred, all_grains, sx, sy)
                    if not remove_edge_grains:
                        all_grains = collect_polygon_from_mask(labels, mask, image_pred, all_grains, sx, sy)

    print(f"Successfully processed {successful_coords}/{len(coords)} coordinates")

    # Post-process (same as existing)
    print("Post-processing...")
    new_grains, comps, g = find_connected_components(all_grains, min_area)
    all_grains = merge_overlapping_polygons(all_grains, new_grains, comps, min_area, image_pred)

    if len(all_grains) > 0:
        labels_out, mask_all = create_labeled_image(all_grains, image)
    else:
        labels_out = np.zeros_like(image[:, :, 0])
        mask_all = np.zeros_like(image)

    if plot_image:
        fig, ax = plt.subplots(figsize=(15, 10))
        ax.imshow(image)
        plot_image_w_colorful_grains(image, all_grains, ax, cmap="Paired")
        plot_grain_axes_and_centroids(all_grains, labels_out, ax, linewidth=1, markersize=10)
        plt.xticks([])
        plt.yticks([])
        plt.xlim([0, np.shape(image)[1]])
        plt.ylim([np.shape(image)[0], 0])
        plt.tight_layout()
    else:
        fig, ax = None, None

    props = regionprops_table(
        labels_out, intensity_image=image,
        properties=("label", "area", "centroid", "major_axis_length",
                   "minor_axis_length", "orientation", "perimeter",
                   "max_intensity", "mean_intensity", "min_intensity"),
    )
    grain_data = pd.DataFrame(props)

    print(f"Pure ONNX Fixed MobileSAM found {len(all_grains)} grains")
    return all_grains, labels_out, mask_all, grain_data, fig, ax
```

Handle Large Images

```python
def predict_large_image(
    fname,
    interpreter,
    encoder_onnx_path,
    predictor_onnx_path,
    min_area,
    patch_size=2000,
    overlap=300,
    remove_large_objects=False,
):
    """
    Predicts the location of grains in a large image using a patch-based approach.

    Parameters
    ----------
    fname : str
        The file path of the input image.
    interpreter : tflite.Interpreter
        The Unet model used for the preliminary grain prediction.
    encoder_onnx_path : str
        The path to the ONNX encoder model.
    predictor_onnx_path : str
        The path to the ONNX predictor model.
    min_area : int
        The minimum area threshold for valid grains.
    patch_size : int, optional
        The size of each patch. Defaults to 2000.
    overlap : int, optional
        The overlap between patches. Defaults to 300.
    remove_large_objects : bool, optional
        Whether to remove large objects from the segmentation. Defaults to False.

    Returns
    -------
    All_Grains : list
        A list of grains represented as polygons.
    image_pred : numpy.ndarray
        The Unet predictions for the entire image.
    all_coords : numpy.ndarray
        The coordinates of the SAM prompts.
    """
    step_size = patch_size - overlap  # step size for overlapping patches
    image = np.array(load_image_for_tflite(fname))
    img_height, img_width = image.shape[:2]  # get the height and width of the image
    # loop over the image and extract patches:
    All_Grains = []
    total_patches = ((img_height - patch_size + step_size) // step_size + 1) * (
        (img_width - patch_size + step_size) // step_size + 1
    )
    # Initialize an array to store the Unet predictions for the entire image
    image_pred = np.zeros((img_height, img_width, 3), dtype=np.float32)

    for i in range(0, img_height - patch_size + step_size + 1, step_size):
        for j in range(0, img_width - patch_size + step_size + 1, step_size):
            patch = image[
                i : min(i + patch_size, img_height), j : min(j + patch_size, img_width)
            ]
            patch_pred = predict_image_tflite(
                patch, interpreter, I=256
            )  # use the Unet model to predict the mask on the patch

            # Define the weights for blending the overlapping regions
            weights = np.ones_like(patch_pred)
            if i > 0:
                weights[:overlap, :] *= np.linspace(0, 1, overlap)[:, None, None]
            if j > 0:
                weights[:, :overlap] *= np.linspace(0, 1, overlap)[None, :, None]
            if i + patch_size < img_height:
                weights[-overlap:, :] *= np.linspace(1, 0, overlap)[:, None, None]
            if j + patch_size < img_width:
                weights[:, -overlap:] *= np.linspace(1, 0, overlap)[None, :, None]

            # Update image_pred with the weighted sum
            image_pred[
                i : min(i + patch_size, img_height), j : min(j + patch_size, img_width)
            ] += (patch_pred * weights)
            labels, coords = seg.label_grains(patch, patch_pred, dbs_max_dist=20.0)
            if (
                len(coords) > 0
            ):  # run the SAM algorithm only if there are grains in the patch
                all_grains, labels, mask_all, grain_data, fig, ax = pure_onnx_fixed_mobilesam_segmentation(
                    encoder_onnx_path=encoder_onnx_path,
                    predictor_onnx_path=predictor_onnx_path,
                    image=patch,
                    image_pred=patch_pred,
                    coords=coords,
                    labels=labels,
                    min_area=min_area,
                    plot_image=True,
                )

                for grain in all_grains:
                    All_Grains += [
                        translate(grain, xoff=j, yoff=i)
                    ]  # translate the grains to the original image coordinates
            patch_num = (
                i // step_size * ((img_width - patch_size + step_size) // step_size + 1)
                + j // step_size
                + 1
            )
            if len(coords) > 0:
                coords[:, 0] = coords[:, 0] + j
                coords[:, 1] = coords[:, 1] + i
                if patch_num > 1:
                    all_coords = np.vstack((all_coords, coords))
                else:
                    all_coords = coords.copy()
            print(f"processed patch #{patch_num} out of {total_patches} patches")
    new_grains, comps, g = seg.find_connected_components(All_Grains, min_area)
    All_Grains = seg.merge_overlapping_polygons(
        All_Grains, new_grains, comps, min_area, patch_pred
    )
    return All_Grains, image_pred, all_coords
```
